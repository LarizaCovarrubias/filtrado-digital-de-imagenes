# 09. Operaciones morfológicas

**Objetivo:** comprender cómo modificar la forma de regiones blancas en imágenes binarias mediante un elemento estructurante.

```python
import numpy as np

from filtrado_digital.visualizacion import mostrar_imagen, comparar
from filtrado_digital.morfologia import (
    erosion_manual,
    dilatacion_manual,
    apertura_manual,
    cierre_manual,
    erosion_opencv,
    dilatacion_opencv,
)
```

## 1. Elemento estructurante

Las operaciones morfológicas analizan principalmente la **forma** de regiones binarias. Utilizan un **elemento estructurante**, una pequeña matriz que indica qué posiciones de la vecindad participan.

En muchas bibliotecas también se le llama *kernel*. Su forma y tamaño determinan qué estructuras pueden conservarse, eliminarse, expandirse o rellenarse.

```python
# Imagen controlada con dos objetos, pequeños huecos y ruido blanco aislado.
img = np.zeros((96, 96), dtype=np.uint8)
img[15:40, 15:40] = 255
img[50:85, 50:85] = 255

img[25, 25] = 0
img[60:63, 60:63] = 0

img[5, 5] = 255
img[90, 10] = 255
img[8, 80] = 255

mostrar_imagen(img, "Imagen binaria de trabajo")
```

## 2. Erosión y dilatación

- **Erosión:** reduce las regiones blancas; objetos pequeños pueden desaparecer.
- **Dilatación:** expande las regiones blancas; separaciones pequeñas pueden reducirse.

```python
ero_m = erosion_manual(img, 3)
dil_m = dilatacion_manual(img, 3)
comparar([img, ero_m, dil_m], ["Original", "Erosión", "Dilatación"])
```

## 3. Apertura y cierre

- **Apertura:** erosión seguida de dilatación. Suele retirar objetos blancos pequeños.
- **Cierre:** dilatación seguida de erosión. Suele rellenar huecos negros pequeños.

El tamaño del elemento estructurante determina qué tan grandes deben ser esas estructuras para verse afectadas.

```python
for tamano in [3, 5]:
    apertura = apertura_manual(img, tamano)
    cierre = cierre_manual(img, tamano)
    comparar(
        [img, apertura, cierre],
        ["Original", f"Apertura {tamano}×{tamano}", f"Cierre {tamano}×{tamano}"],
    )
```

## 4. Implementación manual vs OpenCV

```python
comparar(
    [erosion_manual(img, 3), erosion_opencv(img, 3),
     dilatacion_manual(img, 3), dilatacion_opencv(img, 3)],
    ["Erosión manual", "Erosión OpenCV", "Dilatación manual", "Dilatación OpenCV"],
)
```

## Conclusiones

- La morfología modifica la forma de regiones binarias.
- El elemento estructurante define la vecindad y la escala de la operación.
- Apertura y cierre son útiles para limpiar pequeños objetos o huecos.
- Un elemento estructurante mayor puede eliminar detalles que sí eran importantes.

## Cierre del recorrido

Las técnicas pueden combinarse en un **pipeline**: cargar → ajustar contraste → reducir ruido → detectar/segmentar → limpiar con morfología. Un *pipeline* es una secuencia donde la salida de una etapa se convierte en la entrada de la siguiente.
