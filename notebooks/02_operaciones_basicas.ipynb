# 02. Operaciones básicas

**Objetivo:** aprender a acceder, seleccionar y reorganizar píxeles usando índices y *slicing*, y obtener estadísticas básicas de una imagen.

````python
import numpy as np

from filtrado_digital.io import cargar_imagen, a_grises, ruta_imagen_ejemplo
from filtrado_digital.visualizacion import mostrar_imagen, comparar
from filtrado_digital.transformaciones import (
    reflejo_horizontal_manual,
    reflejo_vertical_manual,
    rotar_90_manual,
    recortar,
    estadisticas_manual,
)
````

## 1. Índices y rebanado (*slicing*)

Un **índice** identifica una posición concreta. En una imagen en grises:

```python
imagen[fila, columna]
```

El **rebanado** o *slicing* selecciona un rango de posiciones sin escribir un ciclo para cada píxel. La forma general es:

```text
inicio : fin : paso
```

El valor `fin` **no se incluye**. Algunos ejemplos:

- `imagen[20:80, 30:100]` selecciona un rectángulo;
- `imagen[:, 40:100]` conserva todas las filas y selecciona columnas;
- `imagen[:, ::-1]` conserva todas las filas y recorre las columnas al revés;
- `imagen[::2, ::2]` toma una fila y columna de cada dos.

````python
matriz = np.arange(30).reshape(5, 6)
print("Matriz completa:\n", matriz)
print("\nFilas 1:4, columnas 2:5:\n", matriz[1:4, 2:5])
print("\nColumnas en orden inverso:\n", matriz[:, ::-1])
````

## 2. Región de trabajo

Usaremos un recorte de la fotografía principal para que las implementaciones con ciclos sean rápidas. El recorte se realiza con *slicing*.

````python
foto = cargar_imagen(ruta_imagen_ejemplo())
imagen = a_grises(foto)
imagen = imagen[180:500, 420:740]
mostrar_imagen(imagen, "Región de trabajo")
````

## 3. Reflejos: ciclos vs NumPy

Un reflejo cambia la orientación de la imagen como si se observara en un espejo. La versión manual ayuda a entender cómo cambian los índices; NumPy permite expresar la misma idea mediante *slicing*.

````python
horizontal_manual = reflejo_horizontal_manual(imagen)
vertical_manual = reflejo_vertical_manual(imagen)

horizontal_numpy = imagen[:, ::-1]
vertical_numpy = imagen[::-1, :]

comparar([imagen, horizontal_manual, vertical_manual], ["Original", "Horizontal", "Vertical"])
print("Horizontal coincide:", np.array_equal(horizontal_manual, horizontal_numpy))
print("Vertical coincide:", np.array_equal(vertical_manual, vertical_numpy))
````

## 4. Rotación de 90°

En una rotación de 90° las filas y columnas intercambian su papel. Compararemos una implementación manual con `np.rot90()`.

````python
rotada_manual = rotar_90_manual(imagen)
rotada_numpy = np.rot90(imagen)
comparar([rotada_manual, rotada_numpy], ["Manual", "NumPy"])
print("Coinciden:", np.array_equal(rotada_manual, rotada_numpy))
````

## 5. Recorte y máscara

Un **recorte** extrae una región rectangular y cambia las dimensiones de la imagen. Una **máscara** conserva normalmente el tamaño original, pero selecciona qué posiciones quedan visibles o participan en una operación.

````python
recorte = recortar(imagen, 60, 240, 80, 260)

mascara = np.zeros_like(imagen)
mascara[70:250, 70:250] = 1
imagen_enmascarada = imagen * mascara

comparar([recorte, imagen_enmascarada], ["Recorte", "Máscara rectangular"])
````

## 6. Estadísticas básicas

La **media** resume la intensidad promedio. La **varianza** mide la dispersión respecto a la media y la **desviación estándar** expresa esa dispersión en las mismas unidades que las intensidades.

````python
manual = estadisticas_manual(imagen)

print("Manual:", manual)
print("NumPy :", {
    "media": float(np.mean(imagen)),
    "varianza": float(np.var(imagen)),
    "desviacion": float(np.std(imagen)),
})
````

## Conclusiones

- Los índices permiten acceder a píxeles concretos.
- El *slicing* permite seleccionar, invertir o saltar regiones de manera compacta.
- Reflejos y rotaciones pueden entenderse como cambios en las posiciones de los píxeles.
- Recortes y máscaras sirven para trabajar con regiones específicas.
