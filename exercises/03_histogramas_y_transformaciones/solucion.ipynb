# Solución 03 — Histogramas y transformaciones

```python
import numpy as np

from filtrado_digital.io import cargar_imagen, a_grises, ruta_imagen_ejemplo
from filtrado_digital.visualizacion import comparar, imagen_e_histograma

imagen = a_grises(cargar_imagen(ruta_imagen_ejemplo()))

float_img = imagen.astype(np.float32)
transformada = np.clip(0.6 * float_img + 40, 0, 255).astype(np.uint8)

comparar([imagen, transformada], ["Original", "s = 0.6r + 40"])
imagen_e_histograma(imagen, "Original")
imagen_e_histograma(transformada, "Transformada")

print("Original:", imagen.min(), imagen.max(), imagen.dtype)
print("Transformada:", transformada.min(), transformada.max(), transformada.dtype)
```

### Interpretación

La multiplicación por `0.6` comprime la separación entre intensidades y el término `+40` desplaza los valores hacia intensidades mayores. Convertir a punto flotante permite hacer la operación antes de limitar y volver a `uint8`.
