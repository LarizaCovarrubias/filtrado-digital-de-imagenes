# Solución 02 — Operaciones básicas

```python
import numpy as np

from filtrado_digital.io import cargar_imagen, a_grises, ruta_imagen_ejemplo
from filtrado_digital.visualizacion import comparar

imagen = a_grises(cargar_imagen(ruta_imagen_ejemplo()))
recorte = imagen[220:380, 440:600]

alto, ancho = recorte.shape
reflejo_manual = np.zeros_like(recorte)
for fila in range(alto):
    for columna in range(ancho):
        reflejo_manual[fila, ancho - 1 - columna] = recorte[fila, columna]

reflejo_numpy = recorte[:, ::-1]

mascara = np.zeros_like(recorte)
mascara[40:120, 40:120] = 1
enmascarada = recorte * mascara

comparar([recorte, reflejo_manual, enmascarada], ["Recorte", "Reflejo manual", "Máscara"])
print("Reflejo coincide:", np.array_equal(reflejo_manual, reflejo_numpy))

suma = 0.0
for valor in recorte.ravel():
    suma += float(valor)
media_manual = suma / recorte.size
print("Media manual:", media_manual)
print("Media NumPy:", float(np.mean(recorte)))
```

### Interpretación

Los ciclos muestran cómo se reasigna cada índice para producir el reflejo. El *slicing* permite expresar la misma inversión con `[:, ::-1]`. La máscara selecciona una región sin cambiar las dimensiones originales.
