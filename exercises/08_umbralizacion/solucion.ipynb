# Solución 08 — Umbralización

```python
import numpy as np

from filtrado_digital.io import cargar_imagen, a_grises, ruta_imagen_ejemplo
from filtrado_digital.thresholding import umbral_global_manual, otsu_manual, umbral_adaptativo_media_manual
from filtrado_digital.visualizacion import comparar

imagen = a_grises(cargar_imagen(ruta_imagen_ejemplo()))[250:442, 480:672]

resultados = [imagen]
titulos = ["Original"]
for t in [80, 120, 160]:
    resultados.append(umbral_global_manual(imagen, t))
    titulos.append(f"T={t}")
comparar(resultados, titulos)

t_otsu, otsu = otsu_manual(imagen)
print("Otsu:", t_otsu)

iluminacion = np.linspace(0.55, 1.15, imagen.shape[1])
no_uniforme = np.clip(imagen * iluminacion[None, :], 0, 255).astype(np.uint8)
global_no_uniforme = umbral_global_manual(no_uniforme, t_otsu)
adapt = umbral_adaptativo_media_manual(no_uniforme, 15, 4)
comparar([no_uniforme, global_no_uniforme, adapt], ["Iluminación desigual", "Global", "Adaptativo"])
```

### Interpretación

El umbral global usa una sola frontera para toda la imagen. Cuando la iluminación varía, regiones del mismo objeto pueden quedar en lados diferentes del umbral. El adaptativo calcula límites locales y puede responder mejor a esas diferencias.
