# Solución 07 — Detección de bordes

```python
import cv2

from filtrado_digital.io import cargar_imagen, a_grises, ruta_imagen_ejemplo
from filtrado_digital.bordes import sobel_opencv, laplaciano_opencv, canny_opencv
from filtrado_digital.visualizacion import comparar

imagen = a_grises(cargar_imagen(ruta_imagen_ejemplo()))[200:392, 440:632]
suave = cv2.GaussianBlur(imagen, (5, 5), 1.0)

_, _, sobel = sobel_opencv(suave)
lap = laplaciano_opencv(suave)
canny_1 = canny_opencv(suave, 40, 100)
canny_2 = canny_opencv(suave, 100, 200)

comparar([suave, sobel, lap, canny_1, canny_2],
         ["Imagen", "Sobel", "Laplaciano", "Canny 40/100", "Canny 100/200"])
```

### Interpretación

Sobel muestra la magnitud de cambios de intensidad; Laplaciano responde fuertemente a variaciones rápidas y puede producir respuestas más sensibles; Canny genera bordes binarios y utiliza dos umbrales. Al aumentar los umbrales, normalmente se conservan menos respuestas y predominan los bordes más fuertes.
