# Solución 05 — CLAHE

```python
import numpy as np
import cv2

from filtrado_digital.io import cargar_imagen, a_grises, ruta_imagen_ejemplo
from filtrado_digital.visualizacion import comparar

imagen = a_grises(cargar_imagen(ruta_imagen_ejemplo()))
iluminacion = np.linspace(0.45, 1.15, imagen.shape[1])
no_uniforme = np.clip(imagen * iluminacion[None, :], 0, 255).astype(np.uint8)

configuraciones = [
    (1.0, (8, 8)),
    (2.0, (8, 8)),
    (2.0, (4, 4)),
    (2.0, (16, 16)),
]

resultados = [no_uniforme]
titulos = ["Entrada"]
for clip, grid in configuraciones:
    clahe = cv2.createCLAHE(clipLimit=clip, tileGridSize=grid)
    resultados.append(clahe.apply(no_uniforme))
    titulos.append(f"clip={clip}, grid={grid}")

comparar(resultados, titulos)
```

### Interpretación

No existe una configuración universalmente mejor: `clipLimit` cambia la intensidad de la mejora local y `tileGridSize` cambia la escala espacial del ajuste. La elección debe basarse en la información que se desea conservar.
