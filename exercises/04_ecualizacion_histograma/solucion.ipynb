# Solución 04 — Ecualización de histograma

```python
import numpy as np
import cv2

from filtrado_digital.io import cargar_imagen, a_grises, ruta_imagen_ejemplo
from filtrado_digital.histogramas import histograma_manual
from filtrado_digital.visualizacion import comparar

imagen = a_grises(cargar_imagen(ruta_imagen_ejemplo()))
imagen = imagen[180:500, 380:700]
bajo = (100 + imagen.astype(np.float32) * (70 / 255)).clip(0, 255).astype(np.uint8)

hist = histograma_manual(bajo)
pdf = hist / bajo.size
cdf = np.cumsum(pdf)
cdf_min = cdf[cdf > 0][0]

lut = np.round((cdf - cdf_min) / (1 - cdf_min) * 255)
lut = np.clip(lut, 0, 255).astype(np.uint8)
manual = lut[bajo]

opencv = cv2.equalizeHist(bajo)
comparar([bajo, manual, opencv], ["Bajo contraste", "Manual", "OpenCV"])
```

### Interpretación

La PDF expresa proporciones, la CDF acumula esas proporciones y la LUT usa la CDF para reasignar las intensidades. La imagen final se obtiene usando la intensidad original como índice de la LUT.
