# Solución 06 — Ruido y filtros de suavizado

```python
from filtrado_digital.io import cargar_imagen, a_grises, ruta_imagen_ejemplo
from filtrado_digital.sinteticas import agregar_ruido_gaussiano, agregar_ruido_sal_pimienta
from filtrado_digital.filtros import (
    filtro_promedio_opencv,
    filtro_gaussiano_opencv,
    filtro_mediana_opencv,
)
from filtrado_digital.visualizacion import comparar

imagen = a_grises(cargar_imagen(ruta_imagen_ejemplo()))[220:412, 500:692]
gauss = agregar_ruido_gaussiano(imagen, sigma=22)
sp = agregar_ruido_sal_pimienta(imagen, proporcion=0.08)

for ruidosa, nombre in [(gauss, "Gaussiano"), (sp, "Sal y pimienta")]:
    comparar(
        [
            ruidosa,
            filtro_promedio_opencv(ruidosa, 5),
            filtro_gaussiano_opencv(ruidosa, 5, 1.2),
            filtro_mediana_opencv(ruidosa, 5),
        ],
        [nombre, "Promedio", "Gaussiano", "Mediana"],
    )

comparar(
    [filtro_gaussiano_opencv(gauss, 3, 1.0), filtro_gaussiano_opencv(gauss, 7, 1.5)],
    ["Gaussiano 3×3", "Gaussiano 7×7"],
)
```

### Interpretación

El filtro de mediana suele manejar especialmente bien los valores extremos del ruido sal-pimienta. Para ruido Gaussiano, el suavizado Gaussiano suele ofrecer un equilibrio útil. Un kernel mayor suaviza más, pero puede perder bordes y detalle fino.
