# Solución 09 — Operaciones morfológicas

```python
import numpy as np

from filtrado_digital.morfologia import apertura_manual, cierre_manual
from filtrado_digital.visualizacion import comparar

imagen = np.zeros((96, 96), dtype=np.uint8)
imagen[15:40, 15:40] = 255
imagen[50:85, 50:85] = 255

imagen[25, 25] = 0
imagen[60:63, 60:63] = 0

imagen[5, 5] = 255
imagen[90, 10] = 255
imagen[8, 80] = 255

for kernel in [3, 5]:
    apertura = apertura_manual(imagen, kernel)
    cierre = cierre_manual(imagen, kernel)
    comparar(
        [imagen, apertura, cierre],
        ["Original", f"Apertura {kernel}×{kernel}", f"Cierre {kernel}×{kernel}"],
    )
```

### Interpretación

La apertura elimina estructuras blancas pequeñas, por eso puede retirar puntos aislados. El cierre rellena huecos negros pequeños. Al aumentar el elemento estructurante, el efecto abarca estructuras mayores, pero también aumenta el riesgo de eliminar o modificar detalles útiles.
