# Solución 01 — Imágenes como matrices

Una posible solución de referencia.

```python
import numpy as np
import matplotlib.pyplot as plt

imagen = np.zeros((200, 200), dtype=np.uint8)
lado = 60
inicio = (imagen.shape[0] - lado) // 2
fin = inicio + lado
imagen[inicio:fin, inicio:fin] = 255

plt.imshow(imagen, cmap="gray", vmin=0, vmax=255)
plt.title("Cuadrado blanco centrado")
plt.axis("off")
plt.show()

contador_blancos = 0
for valor in imagen.ravel():
    if valor == 255:
        contador_blancos += 1

print("shape:", imagen.shape)
print("dtype:", imagen.dtype)
print("mínimo:", imagen.min())
print("máximo:", imagen.max())
print("píxeles blancos:", contador_blancos)
```

### Interpretación

`np.zeros()` crea los **datos** de la imagen como una matriz. `plt.imshow()` toma esa matriz y la representa visualmente. El cuadrado contiene `60 × 60 = 3600` píxeles blancos.
